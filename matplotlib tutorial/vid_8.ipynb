```python
import pandas as pd
from datetime import datetime, timedelta
import matplotlib.pyplot as plt
import matplotlib.dates as mpl_dates
```

```python
plt.style.available
```

```python
plt.style.use('seaborn-v0_8')
```

```python
dates = [
    datetime(2019, 5, 24),
    datetime(2019, 5, 25),
    datetime(2019, 5, 26),
    datetime(2019, 5, 27),
    datetime(2019, 5, 28),
    datetime(2019, 5, 29),
    datetime(2019, 5, 30)
]

y = [0, 1, 3, 4, 6, 5, 7]
```

```python
plt.plot_date(dates, y)

plt.tight_layout()
plt.show()
```

```python
plt.plot(dates, y)

plt.gcf().autofmt_xdate()

plt.tight_layout()
plt.show()
```

```python
plt.plot(dates, y, marker = 'o')

plt.gcf().autofmt_xdate()

date_format = mpl_dates.DateFormatter('%b, %d %Y')

plt.gca().xaxis.set_major_formatter(date_format)

plt.tight_layout()
plt.show()
```

```python
data = pd.read_csv('data_8.csv')

data['Date'] = pd.to_datetime(data['Date'])
data.sort_values('Date', inplace=True)


price_date = data['Date']
price_close = data['Close']
```

```python
plt.plot(price_date, price_close, marker = 'o', linestyle='solid')

plt.gcf().autofmt_xdate()


plt.title('Bitcoin Prices')
plt.xlabel('Date')
plt.ylabel('Closing Price')

plt.tight_layout()
plt.show()
```
