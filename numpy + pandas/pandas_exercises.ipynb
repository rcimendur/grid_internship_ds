#### As two different steps I have to do:
1. Assign it to a variable called data and replace the first 3 columns by a proper datetime index.
2. Year 2061? Do we really have data from this year? Create a function to fix it and apply it.

I will do in a such way that i don't have to declare a new function to fix the year. i will just create a separate DataFrame and include it as a column in the original one and solve two problems at once

```python
df_6 = pd.read_csv(
    'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/06_Stats/Wind_Stats/wind.data',
    sep=r'\s+', 
)

date_cols = pd.DataFrame({
    'year': df_6['Yr'] + 1900,
    'month': df_6['Mo'],
    'day': df_6['Dy']
})
df_6['Yr_Mo_Dy'] = pd.to_datetime(date_cols)
df_6 = df_6.drop(columns = ['Yr', 'Mo', 'Dy'])
df_6.head()
```

```python
df_6.insert(0, 'Yr_Mo_Dy', df_6.pop('Yr_Mo_Dy'))
df_6.head()
```

```python
df_6 = df_6.set_index('Yr_Mo_Dy')
df_6.head()
```

```python
df_6.isnull().sum()
```

```python
df_6.notnull().sum().sum()
```

```python
df_6.sum().sum() / df_6.notna().sum().sum()
```

```python
df_6.describe(percentiles=[])
```

```python
day_stats = pd.DataFrame()

day_stats['min'] = df_6.min(axis = 1)
day_stats['max'] = df_6.max(axis = 1)
day_stats['mean'] = df_6.mean(axis = 1)
day_stats['std'] = df_6.std(axis = 1)

day_stats.head()
```

```python
df_6.loc[df_6.index.month == 1].mean()
```

```python
df_6.resample('YE').mean() # yearly aggregation by avg (end of year)
```

```python
df_6.resample('ME').mean() # monthly aggregation by avg (end of month)
```

```python
df_6.resample('W').mean()  # weekly aggregation by avg
```

```python
weekly = (
    df_6.loc['1961-01-02':].resample('W')
    .agg(['min', 'max', 'mean', 'std'])
    .iloc[:52]
)

# remove the first not full week (starting from 02-01-1961)
# and do the resampling by weeks and get exactly 52 weeks

weekly.head()
```
