#### 29. How to round away from zero a float array ? (★☆☆)


````python
z = np.array([-3.5, -2.1, -0.2, 0.2, 1.4, 4.8])

np.copysign(np.ceil(np.abs(z)), z)
# the idea is to take abs of all elements, round those up to the nearest int, copy the sign of b to magnitude a
````

#### 31. How to ignore all numpy warnings (not recommended)? (★☆☆)

````python
with np.errstate(all='ignore'):
    np.arange(3) / 0 # no warning with context manager

# or 

# global settings change (ignores all warnings for the script)
old_settings = np.seterr(all="ignore")
np.arange(3) / 0 # no warnings globally

# Reset settings back to default
_ = np.seterr(**old_settings)
````

#### 33. How to get the dates of yesterday, today and tomorrow? (★☆☆)

````python
today = np.datetime64('today', 'D')
# today's date using 64-bit datetime representation in day units 'D'

yesterday = today - np.timedelta64(1, 'D')
tomorrow = today + np.timedelta64(1, 'D')

print(today)
print(yesterday)
print(tomorrow)
````

#### 35. How to compute ((A+B)*(-A/2)) in place (without copy)? (★★☆)

````python
# init sample float arrays
A = np.ones(3, dtype=np.float64) * 4
B = np.ones(3, dtype=np.float64) * 2

# step by step operations
np.add(A, B, out=B)
np.divide(A, 2, out=A)
np.negative(A, out=A)
np.multiply(A, B, out=A)

print(A)
````

#### 36. Extract the integer part of a random array of positive numbers using 4 different methods (★★☆)

````python
z = np.random.uniform(0, 10, 10)

# 4 possible methods

m1 = np.floor(z) # truncate by flooring
m2 = np.trunc(z) # truncate in the direction of zeros
m3 = z.astype(int) # drops decimals by changing data type
m4 = z - z % 1 # substract fractional remainder

print(m1, m2, m3, m4, sep='\n')
````

#### 38. Consider a generator function that generates 10 integers and use it to build an array (★☆☆)

````python
def int_generator():
    for x in range(10):
        yield x

# fromiter directly constructs the array from the iterable
z = np.fromiter(int_generator(), dtype = int, count = 10)
print(z)
````

#### 44. Consider a random 10x2 matrix representing cartesian coordinates, convert them to polar coordinates (★★☆)

````python
z = np.random.random((10, 2))
x, y = z[:, 0], z[:, 1]
# generated 10x2 matrix with random points [0;1] and defined the axes
r = np.hypot(x,y) # sqrt(x^2 + y^2) radial distance

theta = np.arctan2(y, x) # theta angle

np.column_stack((r, theta)) #combined
````
