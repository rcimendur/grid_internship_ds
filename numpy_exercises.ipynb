#### 22. Normalize a 5x5 random matrix (★☆☆)

````python
A = np.random.random((5,5))

A_norm = (A - A.min()) / (A.max() - A.min())

print(A_norm)
````

#### 23. Create a custom dtype that describes a color as four unsigned bytes (RGBA) (★☆☆)

````python
# each field is a 8-bit unsigned integer 0 to 255 np.uint8
colors_dtype = np.dtype([
    ('red', np.uint8),
    ('green', np.uint8), 
    ('blue', np.uint8),
    ('alpha', np.uint8)
])

pixel = np.array((255, 255, 0, 255),dtype = colors_dtype)

print(pixel['blue'])
````

#### 24. Multiply a 5x3 matrix by a 3x2 matrix (real matrix product) (★☆☆)

````python
A, B = np.ones((5, 3)), np.ones((3,2))

C = A @ B # or just np.dot(A,B)

print(C)
````

#### 25. Given a 1D array, negate all elements which are between 3 and 8, in place. (★☆☆)

````python
A = np.arange(11)

A[(A > 3) & (A < 8)] *= -1 # condition > 3 AND < 8

print(A)
````

#### 29. How to round away from zero a float array ? (★☆☆)


````python
z = np.array([-3.5, -2.1, -0.2, 0.2, 1.4, 4.8])

np.copysign(np.ceil(np.abs(z)), z)
# the idea is to take abs of all elements, round those up to the nearest int, copy the sign of b to magnitude a
````
